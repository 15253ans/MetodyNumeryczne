## Błąd względny i bezwzględny

***Zadanie 1.***  
Liczbę $\textrm{e}$ możemy zdefiniować jako $\sum\limits_{n=0}^{\infty} \frac{1}{n!}$. Oblicz błąd względny i bezwzględny aproksymacji liczby $\textrm{e}$ poprzez $\textrm{e}^{*}$ w przypadku gdy:  
* $\textrm{e}_1^{*}=\sum\limits_{n=0}^{5} \frac{1}{n!}$    

* $\textrm{e}_2^{*}=\sum\limits_{n=0}^{10} \frac{1}{n!}$  

* $\textrm{e}_3^{*}=\sum\limits_{n=0}^{20} \frac{1}{n!}$  

```python
import math

def calculate_e_star(n_terms):
    return sum(1 / math.factorial(n) for n in range(n_terms + 1))

e_true = math.e
cases = [5, 10, 20]

for n in cases:
    e_star = calculate_e_star(n)
    err_abs = abs(e_true - e_star)
    err_rel = err_abs / abs(e_true)
    print(f"n={n}: e*={e_star:.15f}, Błąd bezwzględny: {err_abs:.2e}, Błąd względny: {err_rel:.2e}")
```

```python
import numpy as np

def analyze_intervals(exponents):
    print(f"{'Wykładnik':>10} | {'Przedział':>20} | {'Długość':>10} | {'Odstęp (step)':>12} | {'Max błąd bezw.'}")
    print("-" * 75)
    
    for exp in exponents:
        start = 2.0**exp
        end = 2.0**(exp + 1)
        interval_length = end - start
        step = interval_length / (2**52)
        max_abs_err = step / 2
        max_rel_err = max_abs_err / start
        print(f"{exp:10} | [{start:g}, {end:g}) | {interval_length:10g} | {step:12.2e} | {max_abs_err:12.2e}")
exponents_to_test = [0, 1, 2, 5, -1, -3]
analyze_intervals(exponents_to_test)

print("\nWnioski:")
print("* W każdym przedziale jest dokładnie tyle samo liczb reprezentowanych: 2^52.")
print("* Odległość między liczbami (krok) zmienia się wraz z wykładnikiem (rośnie dla dużych liczb).")
print("* Maksymalny błąd względny pozostaje stały dla wszystkich przedziałów i wynosi ok. 1.11e-16 (połowa epsilona maszynowego).")
```
